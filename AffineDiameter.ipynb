Note: running on PyPy is much faster.

```python
import math
from itertools import combinations
from collections import defaultdict
from random import sample
from tqdm import tqdm

def minimum_distance(lst, p=127):
    res = float("+inf")
    for c in range(1, p):
        xcs = [x*c % p for x in lst]
        for base in xcs:
            sub = xcs[:]
            for i in range(len(sub)):
                sub[i] -= base
                if sub[i] < 0:
                    sub[i] += p
            res = min(res, max(sub))
    return res

def minimize(lst, p=127):
    v = minimum_distance(lst, p=p)
    for c in range(1, p):
        for base in [x*c % p for x in lst]:
            sub = sorted((x*c - base) % p for x in lst)
            if sub[-1] == v:
                return sub
    raise        
```

# Exhaustive for $3 \le z \le 7$

```python
def exhaustive(z, p=127):
    print("|E| = z =", z)
    
    lst = defaultdict(list)
    total = 0
    seen = 0
    domain = list(range(2, p))
    _total = math.factorial(p) // math.factorial(z-2) // math.factorial(p-(z-2))
    for inds in tqdm(combinations(domain, z-2), total=_total):
        seen += 1
        diam = minimum_distance((0,1) + inds)
        total += diam
        if len(lst[diam]) < 50:
            lst[diam].append(inds)

    global datas
    datas[z] = (lst, seen, total)
    
    print("avg", total / seen)
    for k, v in sorted(dict(lst).items()):
        print(k, len(v), v[:5])
    print("max affine diameter", k, "example", minimize((0, 1) + v[0]))
    print()
    
```

# Random sampling

```python
def random_sampling(z, p=127, HOURS=1):
    lst = defaultdict(list)
    total = 0
    seen = 0
    
    domain = list(range(2, p))
    
    for _ in tqdm(range(int(2*10**7*HOURS))):
        suff = sample(domain, z-2)
        diam = minimum_distance([0,1] + suff)
        total += diam
        seen += 1
        if len(lst[diam]) < 50:
            lst[diam].append(suff)
    
    global datas
    datas[z] = (lst, seen, total)
    
    print("avg", total / seen)
    for k, v in sorted(dict(lst).items()):
        print(k, len(v), v[:5])
    print()
```
